## function to /elimnate space for any data

```python
def strip_all_data(df):
    
   
    df.columns = df.columns.str.strip()


    for col in df.select_dtypes(include=["object", "string"]).columns:
        df[col] = df[col].apply(
            lambda x: x.strip() if isinstance(x, str) else x
        )

    return df
```

## Create a function that can replace a null values for all data types

```python

def replace_nulls(df):
    
    for column in df.columns:
        if df[column].dtype == "str":
            df[column] = df[column].fillna("Unknown")
            
        if df[column].dtype == "str":
            df[column] = df[column].fillna("Unknown")
            
        elif df[column].dtype == "int64":
            df[column] = df[column].fillna(0)
            
        elif df[column].dtype == "float64":
            df[column] = df[column].fillna(0.0)
    
    return df
```

 ## 1. Which manufacturers have the most registered electric vehicles? 

```python
ev_by_make = (
    df.groupby("Make")
    .size()
    .sort_values(ascending=False)
    .head(10)
)

total = len(df)

plt.figure(figsize=(10, 6))

bars = ev_by_make.sort_values().plot(
    kind="barh",
    color="teal"
)

plt.title("Top 10 EV Manufacturers by Registrations")
plt.xlabel("Vehicle Count")
plt.ylabel("Make")

for i, v in enumerate(ev_by_make.sort_values().values):
    percentage = v / total * 100

    bars.text(
        v + ev_by_make.max() * 0.01,
        i,
        f"{percentage:.1f}%",
        va="center",
        fontsize=10
    )

plt.xlim(0, ev_by_make.max() * 1.15)
plt.tight_layout()
plt.show()
```

## 5. What percentage of vehicles are BEVs versus PHEVs?

```python
ev_type = df["Electric Vehicle Type"].value_counts()
ev_type_percentage = (
    df["Electric Vehicle Type"]
      .value_counts(normalize=True)
      .mul(100)
      .round(2)
)
print(ev_type)
print(ev_type_percentage)


def pct_and_count(pct, counts=ev_type.values):
    total = sum(counts)
    val = int(round(pct * total / 100.0))
    return f"{val:,}\n({pct:.1f}%)"

plt.figure(figsize=(7, 7))
plt.pie(
    ev_type_percentage.values,
    labels=ev_type_percentage.index,
    autopct=pct_and_count,
    colors=["cornflowerblue", "salmon"],
    startangle=90
)
plt.title("BEV vs PHEV Share of Registrations")
plt.tight_layout()
plt.show()
```

## 8. BEV vs PHEV mix within the top 5 manufacturers
Business use: Some makers (e.g. Tesla, Nissan) sell almost exclusively BEVs, while others (Kia, Ford, Chevrolet) split their lineup with PHEVs — relevant for charging-infrastructure and incentive planning.

```python
top5 = df["Make"].value_counts().head(5).index


df_top5 = df[
    df["Make"].isin(top5)
    & df["Electric Vehicle Type"].isin([
        "Battery Electric Vehicle (BEV)",
        "Plug-in Hybrid Electric Vehicle (PHEV)"
    ])
]


ev_mix = pd.crosstab(
    df_top5["Make"],
    df_top5["Electric Vehicle Type"]
)


ev_mix = ev_mix.rename(columns={
    "Battery Electric Vehicle (BEV)": "BEV",
    "Plug-in Hybrid Electric Vehicle (PHEV)": "PHEV"
})


ev_pct = ev_mix.div(
    ev_mix.sum(axis=1), axis=0
) * 100


ev_pct = ev_pct.loc[
    ev_mix.sum(axis=1).sort_values(ascending=False).index
]

print(ev_pct.round(2))


ax = ev_pct.plot(
    kind="bar",
    stacked=True,
    figsize=(11, 6),
    color=["seagreen", "orange"]
)


for container in ax.containers:
    labels = [
        f"{bar.get_height():.1f}%"
        if bar.get_height() >= 3 else ""
        for bar in container
    ]
    ax.bar_label(
        container,
        labels=labels,
        label_type="center"
    )

plt.title("BEV vs PHEV Mix: Top 5 Manufacturers")
plt.xlabel("Manufacturer")
plt.ylabel("Percentage of Vehicles (%)")
plt.ylim(0, 100)
plt.xticks(rotation=45, ha="right")
plt.legend(title="Vehicle Type")
plt.tight_layout()
plt.show()
```
